## Idea

```python
n = 2**7
x_lat = qp.Lattice(2,seed=7,order="linear")(n)
x_dn = qp.DigitalNetB2(2,seed=7)(n)
xticks = np.linspace(0,1,251)[:-1]
xmesh0,xmesh1 = np.meshgrid(xticks,xticks)
kshow_lat = qp.KernelShiftInvar(1)(xticks[:,None,None],xticks[None,:,None])
kshow_dn = qp.KernelDigShiftInvar(1,t=10)(xticks[:,None,None],xticks[None,:,None])
n_kmat = 2**4
x_lat_kmat = qp.Lattice(2,seed=7,order="linear")(n_kmat)
x_dn_kmat = qp.DigitalNetB2(2,seed=7)(n_kmat)
kmat_lat = qp.KernelShiftInvar(2,scale=1e3,lengthscales=1)(x_lat_kmat[:,None,:],x_lat_kmat[None,:,:])
kmat_dn = qp.KernelDigShiftInvarCombined(2,scale=1e3,lengthscales=1e-3,t=10)(x_dn_kmat[:,None,:],x_dn_kmat[None,:,:])
fig,ax = pyplot.subplots(nrows=2,ncols=3,figsize=(PW,PW/3*2))
cmap_lat = sns.color_palette("Blues",as_cmap=True) # https://seaborn.pydata.org/tutorial/color_palettes.html
cmap_dn = sns.color_palette("Greens",as_cmap=True) # https://seaborn.pydata.org/tutorial/color_palettes.html
ax[0,0].scatter(x_lat[:,0],x_lat[:,1],color="blue")#cmap_lat.get_over())
ax[1,0].scatter(x_dn[:,0],x_lat[:,1],color="green")#cmap_dn.get_over())
ax[0,1].contourf(xmesh0,xmesh1,kshow_lat,cmap=cmap_lat)
ax[1,1].contourf(xmesh0,xmesh1,kshow_dn,cmap=cmap_dn)
ax[0,2].imshow(kmat_lat,cmap=cmap_lat)
ax[1,2].imshow(kmat_dn,cmap=cmap_dn)
ax[0,0].set_title(r"lattice seq. in $d=2$")
ax[1,0].set_title(r"digital seq. in $d=2$")
ax[0,1].set_title(r"SI kernel $K(x,y)$ in $d=1$")
ax[1,1].set_title(r"DSI kernel $K(x,y)$ in $d=1$")
ax[0,2].set_title(r"Gram matrix SI kernel + lattice seq. ")
ax[1,2].set_title(r"Gram matrix DSI kernel + digital seq. ")
for j in range(2):
    ax[j,2].invert_yaxis()
    ax[j,2].axis("off")
    for k in range(2):
        ax[j,k].set_xlim([0,1])
        ax[j,k].set_xticks([0,1])
        ax[j,k].set_ylim([0,1])
        ax[j,k].set_yticks([0,1])
        ax[j,k].set_aspect(1)
    ax[j,0].set_xlabel(r"$x_1$")
    ax[j,0].set_ylabel(r"$x_2$")
    ax[j,1].set_xlabel(r"$x$")
    ax[j,1].set_ylabel(r"$y$")
fig.savefig("./idea.pdf",bbox_inches="tight")
```

```python
n_max = 2**10 
n_st = 2**4
assert n_max>=n_st
n_mt = [2**6,2**5,2**4]
assert all(n_max>nt for nt in n_mt)
iid = qp.IIDStdUniform(2,seed=5,replications=3)
x_iid = iid(n_max)
ulat = qp.Lattice(2,randomize=False)
x_ulat = ulat(n_max)
shift = np.array([[0.5,0.2],[0.2,0.5],[0.1,0.3]])
x_lat = (shift[:,None,:]+x_ulat)%1
unet = qp.DigitalNetB2(2,randomize=False)
x_unet = unet(n_max)
net = qp.DigitalNetB2(2,seed=3,randomize="DS",replications=3)
x_net = net(n_max)
ncols = 5
fig,ax = pyplot.subplots(nrows=1,ncols=ncols,figsize=(PW,PW/ncols),sharex=True,sharey=True)
ax[0].set_title("IID (multitask) \n")
for t,nt in enumerate(n_mt):
    ax[0].scatter(*x_iid[t,:nt,:].T,color=COLORS[t],marker=MARKERS[t])
ax[1].set_title("lattice seq \n (unrandomized)")
ax[1].scatter(*x_ulat[:n_st,:].T,color="black",marker=MARKERS[0])
ax[1].scatter(*x_lat[2,:n_st,:].T,color=COLORS[2],marker=MARKERS[2])
for i in range(n_st):
    ax[1].quiver(*x_ulat[i,:],*shift[2,:],angles='xy',scale_units='xy',scale=1,width=0.01,color=COLORS[t],alpha=0.25)
    if (x_ulat[i,0]+shift[2,0])>1 or (x_ulat[i,1]+shift[2,1])>1:
        ax[1].quiver(*(x_lat[2,i,:]-shift[2,:]),*shift[2,:],angles='xy',scale_units='xy',scale=1,width=0.01,color=COLORS[t],alpha=0.25)
ax[2].set_title("lattice seqs (multitask) \n  with IID shifts modulo 1")
for t,nt in enumerate(n_mt):
    ax[2].scatter(*x_lat[t,:nt,:].T,color=COLORS[t],marker=MARKERS[t],label=r"task $%d$ with $2^{%d}$ points"%(t,int(np.log2(nt))))
ax[3].set_title("digital seq \n (unrandomized)")
ax[3].scatter(*x_unet[:n_st,:].T,color="black",marker=MARKERS[0])
ax[4].set_title("digital seqs (multitask) \n with IID digital-shifts")
for t,nt in enumerate(n_mt):
    ax[4].scatter(*x_net[t,:nt,:].T,color=COLORS[t],marker=MARKERS[t])
ax[0].set_xlim([0,1]); ax[0].set_xticks([0,1/8,1/4,3/8,1/2,5/8,3/4,7/8,1]); ax[0].set_xticklabels([])#r"$0$","",r"$1/4$","",r"$1/2$","",r"$3/4$","",r"$1$"])
ax[0].set_ylim([0,1]); ax[0].set_yticks([0,1/8,1/4,3/8,1/2,5/8,3/4,7/8,1]); ax[0].set_yticklabels([])#r"$0$","",r"$1/4$","",r"$1/2$","",r"$3/4$","",r"$1$"])
for i in range(ncols):
    for j in range(5):
        ax[i].axhline(y=j/4,color="black",alpha=0.5,linewidth=1)
        ax[i].axvline(x=j/4,color="black",alpha=0.5,linewidth=1)
# for i in range(ncols): ax[i].set_xlabel(r"$x_1$")
# ax[0].set_ylabel(r"$x_2$")
fig.legend(loc="lower center",ncols=len(n_mt),bbox_to_anchor=(1/2,-1/12))
fig.savefig("./points.full.pdf",bbox_inches="tight")
```

```python
def currin(level, x): # https://www.sfu.ca/~ssurjano/curretal88exp.html
    assert x.size(-1)==2 
    assert level in [0,1]
    x1,x2 = x[...,0],x[...,1]
    f = lambda x1,x2: (1-torch.exp(-1/(2*x2)))*(2300*x1**3+1900*x1**2+2092*x1+60)/(100*x1**3+500*x1**2+4*x1+20)
    if level==0:
        y = 1/4*(f(x1+0.05,x2+0.05)+f(x1+0.05,torch.maximum(torch.zeros(1),x2-0.05))+f(x1-0.05,x2+0.05)+f(x1-0.05,torch.maximum(torch.zeros(1),x2-0.05)))
    else:
        y = f(x1,x2)
    return y

def ackley(level, x): # https://www.sfu.ca/~ssurjano/ackley.html
    assert level in [0,1]
    a = 20
    b = 0.2 
    c = 0 if level==0 else 2*np.pi
    t = 2*32.768*x-32.768
    d = x.size(-1)
    y = -a*torch.exp(-b*torch.linalg.norm(t,dim=-1))-torch.exp(1/d*torch.cos(c*t).sum(-1))+a+np.exp(1)
    return y

def park1(level, x): # https://www.sfu.ca/~ssurjano/park91a.html
    assert x.size(-1)==4 
    assert level in [0,1]
    x1,x2,x3,x4 = x[...,0],x[...,1],x[...,2],x[...,3]
    y = x1/2*(torch.sqrt(1+(x2+x3**2)*x4/x1**2)-1)+(x1+3*x4)*torch.exp(1+torch.sin(x3))
    if level==0:
        y = (1+torch.sin(x1)/10)*y-2*x1+x2**2+x3**2+0.5
    return y 

def park2(level, x): # https://www.sfu.ca/~ssurjano/park91b.html
    assert x.size(-1)==4 
    assert level in [0,1]
    x1,x2,x3,x4 = x[...,0],x[...,1],x[...,2],x[...,3]
    y = 2/3*torch.exp(x1+x2)-x4*torch.sin(x3)+x3
    if level==0:
        y = 1.2*y-1
    return y 

def borehole(level, x): # https://www.sfu.ca/~ssurjano/borehole.html
    assert level in [0,1]
    assert x.size(-1)==8
    x = x.numpy()
    import scipy.stats
    rw = torch.from_numpy(scipy.stats.norm(loc=0.10,scale=0.0161812).ppf(x[...,0]))
    r  = torch.from_numpy(scipy.stats.lognorm(scale=np.exp(7.71),s=1.0056).ppf(x[...,1]))
    Tu = torch.from_numpy(scipy.stats.uniform(loc=63070,scale=115600-63070).ppf(x[...,2]))
    Hu = torch.from_numpy(scipy.stats.uniform(loc=990,scale=1110-990).ppf(x[...,3]))
    Tl = torch.from_numpy(scipy.stats.uniform(loc=63.1,scale=116-63.1).ppf(x[...,4]))
    Hl = torch.from_numpy(scipy.stats.uniform(loc=700,scale=820-700).ppf(x[...,5]))
    L  = torch.from_numpy(scipy.stats.uniform(loc=1120,scale=1680-1120).ppf(x[...,6]))
    Kw = torch.from_numpy(scipy.stats.uniform(loc=9855,scale=12045-9855).ppf(x[...,7]))
    C1 = 5 if level==0 else 2
    C2 = 1.5 if level==0 else 1
    frac1 = C1 * torch.pi * Tu * (Hu-Hl)
    frac2a = 2*L*Tu / (torch.log(r/rw)*rw**2*Kw)
    frac2b = Tu / Tl
    frac2 = torch.log(r/rw) * (C2+frac2a+frac2b)
    y = frac1 / frac2
    return y

def elliptic(level, x):
    from scipy.sparse import diags
    from scipy.sparse.linalg import spsolve
    from scipy.stats import norm
    coeffs = x.numpy()
    # Define grid
    N = 2**level  # Number of intervals
    h = 1.0 / N  # Mesh spacing
    x = np.linspace(h, 1 - h, N - 1)  # Interior points
    # Compute diffusion coefficient a(x)
    coeffs = np.random.rand(8) if coeffs is None else coeffs
    assert isinstance(coeffs,np.ndarray)
    coeffs = norm.ppf(coeffs)  # Transform from uniform to iid Gaussian
    batch_shape = list(coeffs.shape)[:-1]
    k = np.arange(1,coeffs.shape[-1]+1)
    a_x = np.exp((coeffs[...,None] / k[:,None]  *np.sin(np.pi * k[:,None] * x)).sum(-2))
    # Compute a at half-grid points (needed for flux terms)
    a_half = np.zeros(batch_shape+[N])
    a_half[...,1:-1] = (a_x[...,:-1] + a_x[...,1:]) / 2  # Midpoint values for flux approximation
    a_half[...,0] = a_x[...,0]  # At the first midpoint
    a_half[...,-1] = a_x[...,-1]  # At the last midpoint
    # Construct the finite difference matrix
    lower_diag = -a_half[...,1:-1] / h**2
    upper_diag = -a_half[...,1:-1] / h**2
    main_diag = (a_half[...,:-1] + a_half[...,1:]) / h**2
    # Right-hand side (forcing term)
    b = np.ones([N-1])  # Constant source term (1)
    def _spsolve(main_diag, upper_diag, lower_diag, b):
        N = main_diag.shape[-1]
        assert main_diag.shape==(N,) and upper_diag.shape==(N-1,) and lower_diag.shape==(N-1,) and b.shape==(N,)
        A = diags([main_diag, upper_diag, lower_diag],[0, 1, -1],shape=[N,N],format="csc",)
        u = spsolve(A, b)
        return u
    vec_spsolve = np.vectorize(_spsolve,signature="(n),(m),(m),(n)->(n)")
    u = vec_spsolve(main_diag, upper_diag, lower_diag, b)
    # Find index closest to x = 0.5
    idx = np.argmin(np.abs(x - 0.5))
    # return u[...,idx], u, x, a_x, x[idx]
    # return torch.from_numpy(u[...,idx])
    return torch.from_numpy(u.max(-1))

def asian_option(level, x):
    assert level>0
    d = 2**level
    assert x.size(-1)>=d
    ao = qp.FinancialOption(
        qp.IIDStdUniform(d),
        option = "ASIAN",
        call_put = "CALL",
        asian_mean = "GEOMETRIC",
        asian_mean_quadrature_rule = "RIGHT",
        volatility = 0.2, 
        start_price = 100, 
        strike_price = 100, 
        interest_rate = 0.05, 
        t_final = 1, 
    )
    y = torch.from_numpy(ao.f(x[...,:d].numpy()))
    return y
```

### Fit GPs Function

```python
def fit_gps(func, 
        d,
        levelops, 
        n,
        iterations, 
        xtest, 
        fit_std_gp_st, 
        fit_std_gp_mt, 
        fit_gpytorch_gp_st, 
        fit_gpytorch_gp_mt, 
        fit_fgpdnet_st, 
        fit_fgpdnet_mt, 
        fit_fgplat_st, 
        fit_fgplat_mt, 
        requires_grad_noise, 
        seeds, 
        rank, 
        sgp_threshold, 
        gpt_threshold,
        verbose,
        predlevels,
        compute_muhats,
        compute_sigmahats,
        n_gp = None,
        n_gpt = None,
        n_fgpdnet = None,
        n_fgplat = None,
        kernel_class_fgpdnet = qp.KernelDigShiftInvarCombined,
        kernel_class_fgplat = qp.KernelShiftInvar
    ):
    levels = len(levelops)
    assert n.shape==(levels,)
    seqs_dnet = [qp.DigitalNetB2(dimension=d,randomize="DS",seed=seeds[l]) for l in range(levels)]
    if fit_std_gp_st or fit_std_gp_mt or fit_gpytorch_gp_st or fit_gpytorch_gp_mt or fit_fgpdnet_st or fit_fgpdnet_mt:
        xs_dnet = [torch.from_numpy(seqs_dnet[l](n[l])) for l in range(levels)]
        ys_dnet = [func(levelops[l],xs_dnet[l]) for l in range(levels)]
    if fit_fgplat_st or fit_fgplat_mt:
        seqs_lat = [qp.Lattice(dimension=d) for l in range(levels)]
        xs_lat = [torch.from_numpy(seqs_lat[l](n[l])) for l in range(levels)]
        # ys_lat = [func(l,xs_lat[l]) for l in range(levels)]
        ys_lat = [func(levelops[l],1-2*torch.abs(xs_lat[l]-1/2)) for l in range(levels)]
    times = []
    yhattests = []
    muhats = []
    sigmahats = []
    if fit_std_gp_st: # single task standard GP 
        if n_gp is not None:
            assert n_gp.shape==n.shape and (n_gp<=n).all()
        else:
            n_gp = n
        if n_gp.sum()>sgp_threshold:
            times.append(np.nan)
            yhattests.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
            if compute_muhats:
                muhats.append(torch.tensor(torch.nan) if np.isscalar(predlevels) else torch.nan*torch.ones(len(predlevels))) 
            if compute_sigmahats:
                sigmahats.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
        else:
            times_std_gp_st = 0
            yhattests_std_gp_st = []
            if compute_sigmahats:
                sigmahats_std_gp_st = []
            if compute_muhats:
                muhats_std_gp_st = []
            if verbose: print("    single task standard GP")
            for l in range(levels):
                if verbose: print("        level %d"%l)
                gp = fastgps.StandardGP(
                    kernel=qp.KernelSquaredExponential(d=d,torchify=True),
                    seqs=seqs_dnet[l],
                    noise = 1e-4,
                    adaptive_nugget = False,
                    requires_grad_noise = requires_grad_noise)
                gp.add_y_next(ys_dnet[l][:n_gp[l]])
                t0 = time.perf_counter()
                data = gp.fit(iterations=iterations,verbose=verbose,store_hists=True,verbose_indent=8)
                times_std_gp_st += (time.perf_counter()-t0)/len(data["iteration"])
                if (np.isscalar(predlevels) and predlevels==l) or ((not np.isscalar(predlevels) and l in predlevels)):
                    yhattests_std_gp_st.append(gp.post_mean(xtest))
                    if compute_muhats:
                        muhats_std_gp_st.append(gp.post_cubature_mean())
                    if compute_sigmahats:
                        sigmahats_std_gp_st.append(torch.sqrt(gp.post_var(xtest)))
            times.append(times_std_gp_st)
            yhattests.append(yhattests_std_gp_st[0] if np.isscalar(predlevels) else torch.stack(yhattests_std_gp_st,dim=0))
            if compute_sigmahats:
                sigmahats.append(sigmahats_std_gp_st[0] if np.isscalar(predlevels) else torch.stack(sigmahats_std_gp_st,dim=0))
            if compute_muhats:
                muhats.append(muhats_std_gp_st[0] if np.isscalar(predlevels) else torch.stack(muhats_std_gp_st,dim=0))
    if fit_std_gp_mt: # multitask standard GP
        if n_gp is not None:
            assert n_gp.shape==n.shape and (n_gp<=n).all()
        else:
            n_gp = n
        if n_gp.sum()>sgp_threshold:
            times.append(np.nan)
            yhattests.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
            if compute_muhats:
                muhats.append(torch.tensor(torch.nan) if np.isscalar(predlevels) else torch.nan*torch.ones(len(predlevels))) 
            if compute_sigmahats:
                sigmahats.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
        else:
            if verbose: print("    multitask standard GP")
            gp = fastgps.StandardGP(
                kernel=qp.KernelMultiTask(
                    base_kernel = qp.KernelSquaredExponential(d=d,torchify=True),
                    num_tasks = levels,
                    rank_factor = levels if rank is None else rank),
                seqs=seqs_dnet,
                noise = 1e-4,
                adaptive_nugget = False,
                requires_grad_noise = requires_grad_noise)
            gp.add_y_next([ys_dnet[l][:n_gp[l]] for l in range(levels)])
            t0 = time.perf_counter()
            data = gp.fit(iterations=iterations,verbose=verbose,store_hists=True)
            times.append((time.perf_counter()-t0)/len(data["iteration"]))
            yhattests.append(gp.post_mean(xtest,task=predlevels))
            if compute_muhats:
                muhats.append(gp.post_cubature_mean(task=predlevels))
            if compute_sigmahats:
                sigmahats.append(torch.sqrt(gp.post_var(xtest,task=predlevels)))
    if fit_gpytorch_gp_st: # single task GPyTorch GP https://docs.gpytorch.ai/en/latest/examples/01_Exact_GPs/Simple_GP_Regression.html
        class ExactGPModel(gpytorch.models.ExactGP):
            def __init__(self, train_x, train_y, likelihood):
                super(ExactGPModel, self).__init__(train_x,train_y,likelihood)
                self.mean_module = gpytorch.means.ConstantMean()
                self.covar_module = gpytorch.kernels.ScaleKernel(gpytorch.kernels.RBFKernel(ard_num_dims=d,))
            def forward(self, x):
                mean_x = self.mean_module(x)
                covar_x = self.covar_module(x)
                return gpytorch.distributions.MultivariateNormal(mean_x,covar_x)
        if n_gpt is not None:
            assert n_gpt.shape==n.shape and (n_gpt<=n).all()
        else:
            n_gpt = n
        if n_gpt.sum()>gpt_threshold:
            times.append(np.nan)
            yhattests.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
            if compute_muhats:
                muhats.append(torch.tensor(torch.nan) if np.isscalar(predlevels) else torch.nan*torch.ones(len(predlevels))) 
            if compute_sigmahats:
                sigmahats.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
        else:
            times_gpytorch_gp_st = 0
            yhattests_gpytorch_gp_st = []
            if compute_sigmahats:
                sigmahats_gpytorch_gp_st = []
            if compute_muhats:
                muhats_gpytorch_gp_st = []
            if verbose: print("    single task GPyTorch GP")
            for l in range(levels):
                if verbose: print("        level %d"%l)
                likelihood = gpytorch.likelihoods.GaussianLikelihood()
                likelihood.noise = 1e-4
                if not requires_grad_noise:
                    likelihood.raw_noise.requires_grad_(False)
                gpt = ExactGPModel(xs_dnet[l][:n_gpt[l]],ys_dnet[l][:n_gpt[l]],likelihood) 
                gpt.train(); likelihood.train()
                optimizer = torch.optim.Rprop(gpt.parameters(),lr=0.1,etas=(0.5,1.2),step_sizes=(0,10))
                mll = gpytorch.mlls.ExactMarginalLogLikelihood(likelihood,gpt)
                t0 = time.perf_counter()
                for i in range(iterations):
                    optimizer.zero_grad()
                    output = gpt(xs_dnet[l][:n_gpt[l]])
                    loss = -mll(output,ys_dnet[l][:n_gpt[l]])
                    if verbose and (i%verbose==0 or i==iterations-1): print("            iter: %-7d GPT loss: %.2e"%(i,loss))
                    loss.backward()
                    optimizer.step()
                times_gpytorch_gp_st += (time.perf_counter()-t0)/iterations
                gpt.eval(); likelihood.eval()
                if (np.isscalar(predlevels) and predlevels==l) or ((not np.isscalar(predlevels) and l in predlevels)):
                    with torch.no_grad(), gpytorch.settings.fast_pred_var():
                        gpt_yhattest_mvngpt = gpt(xtest)
                        gpt_yhattest = gpt_yhattest_mvngpt.mean
                        if compute_sigmahats:
                            sigmahats_gpytorch_gp_st.append(torch.sqrt(gpt_yhattest_mvngpt.variance))
                    yhattests_gpytorch_gp_st.append(gpt_yhattest)
                    if compute_muhats:
                        muhats_gpytorch_gp_st.append(gpt_yhattest.mean(dim=-1)) # cannot exactly compute the mean, so just take a QMC estimate
            times.append(times_gpytorch_gp_st)
            yhattests.append(yhattests_gpytorch_gp_st[0] if np.isscalar(predlevels) else torch.stack(yhattests_gpytorch_gp_st,dim=0))
            if compute_sigmahats:
                sigmahats.append(sigmahats_gpytorch_gp_st[0] if np.isscalar(predlevels) else torch.stack(sigmahats_gpytorch_gp_st,dim=0))
            if compute_muhats:
                muhats.append(muhats_gpytorch_gp_st[0] if np.isscalar(predlevels) else torch.stack(muhats_gpytorch_gp_st,dim=0))
    if fit_gpytorch_gp_mt: # multitask GPyTorch GP https://docs.gpytorch.ai/en/v1.6.0/examples/03_Multitask_Exact_GPs/Hadamard_Multitask_GP_Regression.html
        class MultitaskGPModel(gpytorch.models.ExactGP):
            def __init__(self, train_x, train_y, likelihood):
                super(MultitaskGPModel, self).__init__(train_x, train_y, likelihood)
                self.mean_module = [gpytorch.means.ConstantMean() for i in range(levels)]
                self.covar_module = gpytorch.kernels.ScaleKernel(gpytorch.kernels.RBFKernel(ard_num_dims=d))
                self.task_covar_module = gpytorch.kernels.IndexKernel(num_tasks=levels,rank=levels if rank is None else rank)
            def forward(self,x,i):
                mean_x = torch.stack([self.mean_module[l](x) for l in range(levels)],dim=0)
                mean_x = mean_x[i[:,0],torch.arange(len(i))]
                covar_x = self.covar_module(x)
                covar_i = self.task_covar_module(i)
                covar = covar_x.mul(covar_i)
                return gpytorch.distributions.MultivariateNormal(mean_x,covar)
        if n_gpt is not None:
            assert n_gpt.shape==n.shape and (n_gpt<=n).all()
        else:
            n_gpt = n
        if n_gpt.sum()>gpt_threshold:
            times.append(np.nan)
            yhattests.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
            if compute_muhats:
                muhats.append(torch.tensor(torch.nan) if np.isscalar(predlevels) else torch.nan*torch.ones(len(predlevels))) 
            if compute_sigmahats:
                sigmahats.append(torch.nan*torch.empty(len(xtest)) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
        else:
            if verbose: print("    multitask GPyTorch GP")
            likelihood = gpytorch.likelihoods.GaussianLikelihood()
            likelihood.noise = 1e-4
            if not requires_grad_noise:
                likelihood.noise_covar.raw_noise.requires_grad_(False)
            _full_train_i = torch.cat([l*torch.ones((n_gpt[l],1),dtype=int) for l in range(levels)])
            _full_train_x = torch.cat([xs_dnet[l][:n_gpt[l]] for l in range(levels)])
            _full_train_y = torch.cat([ys_dnet[l][:n_gpt[l]] for l in range(levels)])
            gpt = MultitaskGPModel((_full_train_x,_full_train_i),_full_train_y,likelihood)
            gpt.train(); likelihood.train()
            optimizer = torch.optim.Rprop(gpt.parameters(),lr=0.1,etas=(0.5,1.2),step_sizes=(0,10))
            mll = gpytorch.mlls.ExactMarginalLogLikelihood(likelihood,gpt)
            t0 = time.perf_counter()
            for i in range(iterations):
                optimizer.zero_grad()
                output = gpt(_full_train_x,_full_train_i)
                loss = -mll(output, _full_train_y)
                if verbose and (i%verbose==0 or i==iterations-1): print("        iter: %-7d GPT loss: %.2e"%(i,loss))
                loss.backward()
                optimizer.step()
            times.append((time.perf_counter()-t0)/iterations)
            gpt.eval(); likelihood.eval()
            with torch.no_grad(), gpytorch.settings.fast_pred_var():
                if np.isscalar(predlevels):
                    gpt_yhattest_mvngpt = gpt(xtest,predlevels*torch.ones((xtest.size(0),1),dtype=int))
                    gpt_yhattest = gpt_yhattest_mvngpt.mean
                    if compute_sigmahats:
                        sigmahats.append(torch.sqrt(gpt_yhattest_mvngpt.variance))
                else:
                    gpt_yhattest_mvns = [gpt(xtest,pl*torch.ones((xtest.size(0),1),dtype=int)) for pl in predlevels]
                    gpt_yhattest = torch.stack([mvn.mean for mvn in gpt_yhattest_mvns],axis=0)
                    if compute_sigmahats:
                        sigmahats.append(torch.sqrt(torch.stack([mvn.variance for mvn in gpt_yhattest_mvns],axis=0)))
            yhattests.append(gpt_yhattest)
            if compute_muhats:
                muhats.append(gpt_yhattest.mean(dim=-1)) # cannot exactly compute the mean, so just take a QMC estimate
    if fit_fgpdnet_st: # single task fast GP Dnet
        if n_fgpdnet is not None:
            assert n_fgpdnet.shape==n.shape and (n_fgpdnet<=n).all()
        else:
            n_fgpdnet = n
        times_fgp_dnet_st = 0
        yhattests_fgp_dnet_st = []
        if compute_sigmahats:
            sigmahats_fgp_dnet_st = []
        if compute_muhats:
            muhats_fgp_dnet_st = []
        if verbose: print("    single task fast GP Dnet")
        for l in range(levels):
            if verbose: print("        level %d"%l)
            fgp_dnet = fastgps.FastGPDigitalNetB2(
                kernel=kernel_class_fgpdnet(d=d,torchify=True),
                seqs=seqs_dnet[l],
                noise = 1e-4 if requires_grad_noise else 2*qp.util.transforms.EPS64,
                adaptive_nugget = False,
                requires_grad_noise = requires_grad_noise)
            fgp_dnet.add_y_next(ys_dnet[l][:n_fgpdnet[l]])
            t0 = time.perf_counter()
            data = fgp_dnet.fit(iterations=iterations,verbose=verbose,store_hists=True,verbose_indent=8)
            times_fgp_dnet_st += (time.perf_counter()-t0)/len(data["iteration"])
            if (np.isscalar(predlevels) and predlevels==l) or ((not np.isscalar(predlevels) and l in predlevels)):
                yhattests_fgp_dnet_st.append(fgp_dnet.post_mean(xtest))
                if compute_muhats:
                    muhats_fgp_dnet_st.append(fgp_dnet.post_cubature_mean())
                if compute_sigmahats:
                    sigmahats_fgp_dnet_st.append(torch.sqrt(fgp_dnet.post_var(xtest)))
        times.append(times_fgp_dnet_st)
        yhattests.append(yhattests_fgp_dnet_st[0] if np.isscalar(predlevels) else torch.stack(yhattests_fgp_dnet_st,dim=0))
        if compute_sigmahats:
            sigmahats.append(sigmahats_fgp_dnet_st[0] if np.isscalar(predlevels) else torch.stack(sigmahats_fgp_dnet_st,dim=0))
        if compute_muhats:
            muhats.append(muhats_fgp_dnet_st[0] if np.isscalar(predlevels) else torch.stack(muhats_fgp_dnet_st,dim=0))
    if fit_fgpdnet_mt: # multitask fast GP Dnet 
        if n_fgpdnet is not None:
            assert n_fgpdnet.shape==n.shape and (n_fgpdnet<=n).all()
        else:
            n_fgpdnet = n
        if verbose: print("    multitask fast GP Dnet")
        fgp_dnet = fastgps.FastGPDigitalNetB2(
            kernel = qp.KernelMultiTask(
                base_kernel = kernel_class_fgpdnet(d=d,torchify=True),
                num_tasks = levels,
                rank_factor = levels if rank is None else rank,),
            seqs=seqs_dnet,
            noise = 1e-4 if requires_grad_noise else 2*qp.util.transforms.EPS64,
            adaptive_nugget = False,
            requires_grad_noise = requires_grad_noise,)
        fgp_dnet.add_y_next([ys_dnet[l][:n_fgpdnet[l]] for l in range(levels)])
        t0 = time.perf_counter()
        data = fgp_dnet.fit(iterations=iterations,verbose=verbose,store_hists=True)
        times.append((time.perf_counter()-t0)/len(data["iteration"]))
        yhattests.append(fgp_dnet.post_mean(xtest,task=predlevels))
        if compute_muhats:
            muhats.append(fgp_dnet.post_cubature_mean(task=predlevels))
        if compute_sigmahats:
            sigmahats.append(torch.sqrt(fgp_dnet.post_var(xtest,task=predlevels)))
    if fit_fgplat_st: # single task fast GP lattice
        if n_fgplat is not None:
            assert n_fgplat.shape==n.shape and (n_fgplat<=n).all()
        else:
            n_fgplat = n
        times_fgp_lat_st = 0
        yhattests_fgp_lat_st = []
        if compute_sigmahats:
            sigmahats_fgp_lat_st = []
        if compute_muhats:
            muhats_fgp_lat_st = []
        if verbose: print("    single task fast GP lattice")
        for l in range(levels):
            if verbose: print("        level %d"%l)
            fgp_lat = fastgps.FastGPLattice(
                kernel=kernel_class_fgplat(d=d,torchify=True),
                seqs=seqs_lat[l],
                noise = 1e-4 if requires_grad_noise else 2*qp.util.transforms.EPS64,
                adaptive_nugget = False,
                requires_grad_noise = requires_grad_noise)
            fgp_lat.add_y_next(ys_lat[l][:n_fgplat[l]])
            t0 = time.perf_counter()
            data = fgp_lat.fit(iterations=iterations,verbose=verbose,store_hists=True,verbose_indent=8)
            times_fgp_lat_st += (time.perf_counter()-t0)/len(data["iteration"])
            if (np.isscalar(predlevels) and predlevels==l) or ((not np.isscalar(predlevels) and l in predlevels)):
                yhattests_fgp_lat_st.append(1/2*(fgp_lat.post_mean(xtest/2)+fgp_lat.post_mean(1-xtest/2)))
                if compute_muhats:
                    muhats_fgp_lat_st.append(fgp_lat.post_cubature_mean())
                if compute_sigmahats:
                    sigmahats_fgp_lat_st.append(torch.sqrt(fgp_lat.post_var(xtest)))
        times.append(times_fgp_lat_st)
        yhattests.append(yhattests_fgp_lat_st[0] if np.isscalar(predlevels) else torch.stack(yhattests_fgp_lat_st,dim=0))
        if compute_sigmahats:
            sigmahats.append(sigmahats_fgp_lat_st[0] if np.isscalar(predlevels) else torch.stack(sigmahats_fgp_lat_st,dim=0))
        if compute_muhats:
            muhats.append(muhats_fgp_lat_st[0] if np.isscalar(predlevels) else torch.stack(muhats_fgp_lat_st,dim=0))
    if fit_fgplat_mt: # multitask fast GP lattice
        if n_fgplat is not None:
            assert n_fgplat.shape==n.shape and (n_fgplat<=n).all()
        else:
            n_fgplat = n
        if verbose: print("    multitask fast GP lattice")
        fgp_lat = fastgps.FastGPLattice(
            kernel = qp.KernelMultiTask(
                base_kernel = kernel_class_fgplat(d=d,torchify=True,alpha=2),
                num_tasks = levels,
                rank_factor = levels if rank is None else rank),
            seqs = seqs_lat,
            noise = 1e-4 if requires_grad_noise else 2*qp.util.transforms.EPS64,
            adaptive_nugget = False,
            requires_grad_noise = requires_grad_noise,
            )
        fgp_lat.add_y_next([ys_lat[l][:n_fgplat[l]] for l in range(levels)])
        t0 = time.perf_counter()
        data = fgp_lat.fit(iterations=iterations,verbose=verbose,store_hists=True)
        times.append((time.perf_counter()-t0)/len(data["iteration"]))
        yhattests.append(1/2*(fgp_lat.post_mean(xtest/2,task=predlevels)+fgp_lat.post_mean(1-xtest/2,task=predlevels)))
        if compute_muhats:
            muhats.append(fgp_lat.post_cubature_mean(task=predlevels))
        if compute_sigmahats:
            sigmahats.append(torch.nan*torch.empty(torch.nan*torch.empty(len(xtest))) if np.isscalar(predlevels) else torch.nan*torch.empty((len(predlevels),len(xtest))))
    return torch.tensor(times),torch.stack(yhattests,dim=0),(torch.stack(muhats,dim=0) if compute_muhats else None),(torch.stack(sigmahats,dim=0) if compute_sigmahats else None)
```

### Ackley Example

```python
def ackley3(level, x): # https://www.sfu.ca/~ssurjano/ackley.html
    assert level in [0,1,2]
    a = 20
    b = 0.2 
    if level==0:
        c = 0 
    elif level==1:
        c = np.pi/2
    elif level==2:
        c = 2*np.pi
    t = 2*32.768*x-32.768
    d = x.size(-1)
    y = -a*torch.exp(-b*torch.linalg.norm(t,dim=-1))-torch.exp(1/d*torch.cos(c*t).sum(-1))+a+np.exp(1)
    return y
ALPHA = 0.25
seeds = [11,12,13]
ns = [32,16,8]
xticks = torch.linspace(0,1,100)[1:-1,None]
yticks = torch.stack([ackley3(l,xticks) for l in range(3)],dim=0)
times,yhattests,muhats,sigmahats = fit_gps(
    func = ackley3,
    d = 1,
    levelops = [0,1,2],
    n = torch.tensor(ns,dtype=int),
    iterations = 100,
    xtest = xticks,
    fit_std_gp_st = True,
    fit_std_gp_mt = True,
    fit_gpytorch_gp_st = True,
    fit_gpytorch_gp_mt = True,
    fit_fgpdnet_st = True,
    fit_fgpdnet_mt = True,
    fit_fgplat_st = True, 
    fit_fgplat_mt = True, 
    requires_grad_noise = False,
    seeds = seeds,
    rank = 3,
    sgp_threshold = 2**10, 
    gpt_threshold = 2**10,
    verbose = 25, 
    predlevels = [0,1,2],
    compute_muhats = True, 
    compute_sigmahats = True, 
)
xs = [torch.from_numpy(qp.DigitalNetB2(1,seed=seeds[l],randomize="DS")(ns[l])) for l in range(3)]
ys = [ackley3(l,xs[l]) for l in range(3)]
print(times.shape)
print(yhattests.shape)
print(muhats.shape)
print(sigmahats.shape)
```

### Run Function

```python
def run(
        func, 
        d, 
        levelops, 
        ns, 
        iterations = 100, 
        trials = 1, 
        fit_std_gp_st = True, 
        fit_std_gp_mt = True, 
        fit_gpytorch_gp_st = True, 
        fit_gpytorch_gp_mt = True, 
        fit_fgpdnet_st = True, 
        fit_fgpdnet_mt = True, 
        fit_fgplat_st = True, 
        fit_fgplat_mt = True, 
        requires_grad_noise = False, 
        n_ref = 2**18, 
        n_test = 2**10, 
        ref_mu = None, 
        seeds = None, 
        test_data = None, 
        tf_output = lambda x: x,
        rank = None,
        sgp_threshold = 2**11,
        gpt_threshold = 2**12,
        verbose = False):
    levels = len(levelops)
    assert ns.ndim==2 and ns.size(1)==levels
    assert n_ref>=n_test
    num_ns = len(ns)
    num_gptypes = int(fit_std_gp_st)+int(fit_gpytorch_gp_st)+int(fit_fgpdnet_st)+int(fit_fgplat_st)+int(fit_std_gp_mt)+int(fit_gpytorch_gp_mt)+int(fit_fgpdnet_mt)+int(fit_fgplat_mt)
    gptypes = []
    if fit_std_gp_st: gptypes.append("Indep. Cholesky SE")
    if fit_std_gp_mt: gptypes.append("Multitask Cholesky SE")
    if fit_gpytorch_gp_st: gptypes.append("Indep. CG SE (GPyTorch)")
    if fit_gpytorch_gp_mt: gptypes.append("Multitask CG SE (GPyTorch)")
    if fit_fgpdnet_st: gptypes.append("Indep. Fast DSI (ours)")
    if fit_fgpdnet_mt: gptypes.append("Multitask Fast DSI (ours)")
    if fit_fgplat_st: gptypes.append("Indep. Fast SI Lat (ours)")
    if fit_fgplat_mt: gptypes.append("Multitask Fast SI Lat (ours)")
    if test_data is None:
        xtest = torch.from_numpy(qp.Halton(dimension=d,randomize="QRNG",seed=11)(n_ref))
        ytest = func(levelops[-1],xtest)
    else:
        xtest,ytest = test_data
    if ref_mu is None:
        mu = ytest.mean() # true mean computed with large-sample QMC estimate
    else:
        mu = ref_mu
    times = torch.empty((num_gptypes,num_ns,trials))
    l2rerrors = torch.empty((num_gptypes,num_ns,trials))
    bcerrors = torch.empty((num_gptypes,num_ns,trials))
    if seeds is None: seeds = [None]*levels
    print("\tgptypes:",gptypes)
    print()
    print("\tns")
    for i in range(num_ns): print("\t\t%d: %s"%(i,ns[i].numpy()))
    print()
    print("\t~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~")
    print()
    t0 = time.perf_counter()
    for i in range(num_ns):
        print("\tns[%d] = %s"%(i,ns[i].numpy()))
        print("\t\t%-10s %-20s %-35s %-35s %-35s"%("trial","clock","times","l2rerrors","bcerrors"))
        for j in range(trials):
            times[:,i,j],yhattest,muhat,_ = fit_gps(
                func=func,
                d=d,
                levelops=levelops,
                n=ns[i],
                iterations=iterations,
                xtest=xtest[:n_test],
                fit_std_gp_st=fit_std_gp_st,
                fit_std_gp_mt=fit_std_gp_mt,
                fit_gpytorch_gp_st=fit_gpytorch_gp_st,
                fit_gpytorch_gp_mt=fit_gpytorch_gp_mt,
                fit_fgpdnet_st=fit_fgpdnet_st,
                fit_fgpdnet_mt=fit_fgpdnet_mt,
                fit_fgplat_st=fit_fgplat_st,
                fit_fgplat_mt=fit_fgplat_mt,
                requires_grad_noise=requires_grad_noise,
                seeds=seeds,
                rank=rank,
                sgp_threshold=sgp_threshold,
                gpt_threshold=gpt_threshold,
                verbose=verbose,
                predlevels=levels-1,
                compute_muhats=True,
                compute_sigmahats=False,
                )
            yhattest = tf_output(yhattest)
            l2rerrors[:,i,j] = torch.linalg.norm(yhattest-ytest[:n_test],dim=-1)/torch.linalg.norm(ytest[:n_test])
            bcerrors[:,i,j] = torch.abs(muhat-mu)
            with np.printoptions(formatter={"float":lambda x: "%.1e"%x}):
                print("\t\t%-10d %-20.1f %-35s %-35s %-35s"%(j,time.perf_counter()-t0,times[:,i,j].numpy(),l2rerrors[:,i,j].numpy(),bcerrors[:,i,j].numpy()))
        print()
    return gptypes,times,l2rerrors,bcerrors
```

### MLP: Normalization

```python
assert (y0_train_qgp.argmin(0)==0).all()
assert (y1_train_qgp.argmin(0)==0).all()
assert (y2_train_qgp.argmin(0)==0).all()
assert (y0_train_qgp.argmax(0)==44).all()
assert (y1_train_qgp.argmax(0)==44).all()
assert (y2_train_qgp.argmax(0)==44).all()
assert (y0_test_qgp.argmin(0)==0).all()
assert (y1_test_qgp.argmin(0)==0).all()
assert (y2_test_qgp.argmin(0)==0).all()
assert (y0_test_qgp.argmax(0)==44).all()
assert (y1_test_qgp.argmax(0)==44).all()
assert (y2_test_qgp.argmax(0)==44).all()
class MLPNormalization(torch.nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(MLPNormalization,self).__init__()
        self.fc1 = torch.nn.Linear(input_size,hidden_size)
        self.activation1 = torch.nn.Tanh()
        self.fc2 = torch.nn.Linear(hidden_size,hidden_size)
        self.activation2 = torch.nn.Tanh()
        self.fc3 = torch.nn.Linear(hidden_size,hidden_size)
        self.activation3 = torch.nn.Tanh()
        self.fc4 = torch.nn.Linear(hidden_size,hidden_size)
        self.activation4 = torch.nn.Tanh()
        self.fc5 = torch.nn.Linear(hidden_size,output_size)
    def forward(self, theta):
        theta = self.fc1(theta)
        theta = self.activation1(theta)
        theta = self.fc2(theta)
        theta = self.activation2(theta)
        theta = self.fc3(theta)
        theta = self.activation3(theta)
        theta = self.fc4(theta)
        theta = self.activation4(theta)
        theta = self.fc5(theta)
        return torch.abs(theta)
def init_weights(m):
    if isinstance(m,torch.nn.Linear):
        torch.nn.init.xavier_uniform_(m.weight)
        if m.bias is not None:
            torch.nn.init.zeros_(m.bias)
model_normalization = MLPNormalization(input_size=11,hidden_size=11,output_size=1)
torch.manual_seed(11)
model_normalization.apply(init_weights)
# optimizer = torch.optim.Adam(model.parameters(),lr=1e-2)
criterion = torch.nn.MSELoss()
optimizer = torch.optim.LBFGS(model_normalization.parameters(),lr=1e-2)
epochs = 5000
x_train_mlp = torch.cat([
    torch.cat([x0_train,y0_train_qgp[0,:,None]],dim=-1),
    torch.cat([x1_train,y1_train_qgp[0,:,None]],dim=-1),
    torch.cat([x2_train,y2_train_qgp[0,:,None]],dim=-1)],dim=0)
y_train_mlp = torch.cat([y0_train_qgp,y1_train_qgp,y2_train_qgp],dim=-1)
x_test_mlp = torch.cat([
    torch.cat([x0_test,y0_test_qgp[0,:,None]],dim=-1),
    torch.cat([x1_test,y1_test_qgp[0,:,None]],dim=-1),
    torch.cat([x2_test,y2_test_qgp[0,:,None]],dim=-1)],dim=0)
y_test_mlp = torch.cat([y0_test_qgp,y1_test_qgp,y2_test_qgp],dim=-1)
for epoch in range(epochs):
    def closure():
        optimizer.zero_grad()
        loss = criterion(model_normalization(x_train_mlp)[:,0]+torch.log(y_train_mlp[0]),torch.log(y_train_mlp[-1]))
        loss.backward()
        return loss 
    optimizer.step(closure)
    with torch.no_grad():
        current_train_loss = criterion(model_normalization(x_train_mlp)[:,0]+torch.log(y_train_mlp[0]),torch.log(y_train_mlp[-1]))
        current_test_loss = criterion(model_normalization(x_test_mlp)[:,0]+torch.log(y_test_mlp[0]),torch.log(y_test_mlp[-1]))
    if epoch==0 or (epoch+1)%(epochs/20)==0:
        print(f'Epoch [{epoch+1}/{epochs}], train loss: {current_train_loss.item():.3e}, test loss: {current_test_loss.item():.3e}')
norm_log_const_train = model_normalization(x_train_mlp)[:,0].detach()
norm_log_const_test = model_normalization(x_test_mlp)[:,0].detach()
y_train_qgp_normed = torch.exp((torch.log(y_train_mlp)-torch.log(y_train_mlp[[0]]))/norm_log_const_train)
y_test_qgp_normed = torch.exp((torch.log(y_test_mlp)-torch.log(y_test_mlp[[0]]))/norm_log_const_test)
y0_train_qgp_normed = y_train_qgp_normed[:,:n_train[0]]
y1_train_qgp_normed = y_train_qgp_normed[:,n_train[0]:-n_train[2]]
y2_train_qgp_normed = y_train_qgp_normed[:,-n_train[2]:]
y0_test_qgp_normed = y_test_qgp_normed[:,:n_test[0]]
y1_test_qgp_normed = y_test_qgp_normed[:,n_test[0]:-n_test[2]]
y2_test_qgp_normed = y_test_qgp_normed[:,-n_test[2]:]
keep_y0 = (y0_test_qgp_normed<y_train_qgp_normed.amax()).all(0)
keep_y1 = (y1_test_qgp_normed<y_train_qgp_normed.amax()).all(0)
keep_y2 = (y2_test_qgp_normed<y_train_qgp_normed.amax()).all(0)
x0_test_normed = x0_test[keep_y0]
x1_test_normed = x1_test[keep_y1]
x2_test_normed = x2_test[keep_y2]
y0_test_qgp_normed = y0_test_qgp_normed[:,keep_y0]
y1_test_qgp_normed = y1_test_qgp_normed[:,keep_y1]
y2_test_qgp_normed = y2_test_qgp_normed[:,keep_y2]
y_test_qgp_normed = torch.cat([y0_test_qgp_normed,y1_test_qgp_normed,y2_test_qgp_normed],dim=1)
n_test_normed = torch.tensor([x0_test_normed.size(0),x1_test_normed.size(0),x2_test_normed.size(0)],dtype=int)
print("x0_test_normed.shape = %s"%str(tuple(x0_test_normed.shape)))
print("x1_test_normed.shape = %s"%str(tuple(x1_test_normed.shape)))
print("x2_test_normed.shape = %s"%str(tuple(x2_test_normed.shape)))
print("y0_test_qgp_normed.shape = %s"%str(tuple(y0_test_qgp_normed.shape)))
print("y1_test_qgp_normed.shape = %s"%str(tuple(y1_test_qgp_normed.shape)))
print("y2_test_qgp_normed.shape = %s"%str(tuple(y2_test_qgp_normed.shape)))
nrows = 2
ncols = 10
fig,ax = pyplot.subplots(nrows=nrows,ncols=ncols,figsize=(2*PW,2*PW/ncols*nrows),sharex=False,sharey=False)
for j in range(ncols):
    ax[0,j].plot(energy_levels,y_train_qgp_normed[:,j],'-o',color=COLORS[0])
    ax[1,j].plot(energy_levels,y_test_qgp_normed[:,j],'-o',color=COLORS[1])
    ax[0,j].set_title("y_train_qgp_normed[:,%d]"%j)
    ax[1,j].set_title("y_test_qgp_normed[:,%d]"%j)
    for i in range(nrows):
        pass
        ax[i,j].set_xscale("log",base=10)
        # ax[i,j].set_yscale("log",base=10)
fig.tight_layout()
print("y_train_qgp_normed.amax() = %.2f"%y_train_qgp_normed.amax())
print("y_test_qgp_normed.amax() = %.2f"%y_test_qgp_normed.amax())
fig,ax = pyplot.subplots(nrows=1,ncols=2,figsize=(PW/2,PW/4))
ax[0].hist(y_train_qgp_normed.amax(0),bins=16,density=True)
ax[1].hist(y_test_qgp_normed.amax(0),bins=16,density=True)
ax[0].set_title("max y_train_qgp_normed")
ax[1].set_title("max y_test_qgp_normed")
fig.tight_layout()
print("y_train_qgp_normed.shape = %s"%str(tuple(y_train_qgp_normed.shape)))
print("y_test_qgp_normed.shape = %s"%str(tuple(y_test_qgp_normed.shape)))
```

### GP Fitting

```python
y0_train_qoi = y0_train_qgp_normed[25]#.mean(0)
y1_train_qoi = y1_train_qgp_normed[25]#.mean(0)
y2_train_qoi = y2_train_qgp_normed[25]#.mean(0)
y0_test_qoi = y0_test_qgp_normed[25]#.mean(0)
y1_test_qoi = y1_test_qgp_normed[25]#.mean(0)
y2_test_qoi = y2_test_qgp_normed[25]#.mean(0)
xtest_qoi = torch.cat([x0_test_normed[:,1:],x1_test_normed[:,1:],x2_test_normed[:,1:]],dim=0)
```

```python
def func_qgp(level, x): # https://www.sfu.ca/~ssurjano/ackley.html
    assert level in [0,1,2]
    if level==0:
        assert torch.allclose(x,x0_train[:,1:])
        return y0_train_qoi
    elif level==1:
        assert torch.allclose(x,x1_train[:,1:])
        return y1_train_qoi
    else:
        assert torch.allclose(x,x2_train[:,1:])
        return y2_train_qoi
times,yhattests,_,_ = fit_gps(
    func = func_qgp,
    d = d,
    levelops = [0,1,2],
    n = n_train,
    iterations = 500,
    xtest = xtest_qoi,
    fit_std_gp_st = True,
    fit_std_gp_mt = True,
    fit_gpytorch_gp_st = True,
    fit_gpytorch_gp_mt = True,
    fit_fgpdnet_st = True,
    fit_fgpdnet_mt = True,
    fit_fgplat_st = False, 
    fit_fgplat_mt = False, 
    requires_grad_noise = False,
    seeds = seeds,
    rank = 3,
    sgp_threshold = 2**14, 
    gpt_threshold = 2**14,
    verbose = 25, 
    predlevels = [0,1,2],
    compute_muhats = False, 
    compute_sigmahats = False, 
)
torch.save(times,"./qgp_metrics/times.pt")
torch.save(yhattests,"./qgp_metrics/yhattests.pt")
```
